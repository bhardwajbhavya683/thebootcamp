```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text = "For any true **bibliophile** or self-proclaimed **bookworm**, entering the world of stories is like unlocking infinite parallel universes. Whether you are getting lost in epic **novels**, quick-paced **novellas**, or deeply detailed **fan-fiction**, the act of **reading** transforms the quietest room into an avenue for endless exploration. Creative **authors** and master **writers** build extraordinary realms across every conceivable genre—from the mind-bending possibilities of **science-fiction**, adrenaline-fueled **action**, and high-stakes **adventure** to spine-chilling **horror** and heart-pounding **thrillers**. Beyond classic **fiction**, millions of modern readers flip pages of vibrant **manga** and **manhwa**, immersing themselves in diverse narratives ranging from sweet **romance**, hilarious **rom-com** **comedy**, and thrilling **yaoi** and **yuri** dynamics. With a stash of trusty **bookmarks** marking saved places across stacked nightstands, every dedicated **reader** knows there is no greater joy than turning the final page of a beloved story—only to immediately reach for the next one."
print(text)
```

```python
words = word_tokenize(text.lower())
print(words)
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
print(filtered_words)
```

```python
word_freq = Counter(filtered_words)
print(word_freq)
```

```python
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
wordcloud.to_file("my_wordcloud.jpg")
```

```python
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
