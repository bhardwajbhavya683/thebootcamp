```python
import random

# Generate a list of 15 random numbers between 1 and 100
random_numbers = [random.randint(1, 100) for _ in range(15)]
print("Original List:", random_numbers)

# Filter and print even numbers
even_numbers = [num for num in random_numbers if num % 2 == 0]
print("Even Numbers:", even_numbers)
```

```python
# Sample string
text = "Hello Bootcamp World"
print("Original String:", text)

# 1. Reverse the entire string character-by-character
reversed_string = text[::-1]
print("Reversed String:", reversed_string)

# 2. Reverse the word order
reversed_words = " ".join(text.split()[::-1])
print("Reversed Words Order:", reversed_words)
```
